```python
import numpy as np
import pandas as pd
```

```python
emp = pd.DataFrame({
    'epid' : [1, 2, 3, 4, 5],
    'name' : ['v', 'k', 'h', 't', 'y'],
    'dep' : ['HR', 'It', 'HPP', 'MNN', 'WEE']
})

asll = pd.DataFrame({
    'epid' : [1, 2, 3, 6, 7],
    'sal' : [45, 453, 4634, 56, 456],
    'bonus' : [44536, 6453, 436, 56, 67]
})
```

```python
emp
```

```python
asll
```

```python
pd.merge(emp, asll, on='epid')
```

```python
pd.merge(emp, asll, on='epid', how="outer")
```

```python
pd.merge(emp, asll, on='epid', how= "left")
```

```python
pd.merge(emp, asll, on='epid', how= "right")
```
