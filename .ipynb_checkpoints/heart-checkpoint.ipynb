```python
numeric_col = df[['Age', 'RestingBP',	'Cholesterol',	'FastingBS',	'MaxHR',	'Oldpeak',	'HeartDisease']]
```

```python
numeric_col
```

```python
for col in numeric_col:
    plt.figure(figsize=(6,5))
    sns.histplot(df[col], kde = True, bins=20)
```

```python
sns.countplot
```

```python
for col in numeric_col:
    plt.figure(figsize=(6,4))
    sns.boxplot(x = df[col])
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df.corr(numeric_only=True), annot = True)
```

```python
ch_mean = df.loc[df['Cholesterol'] !=0, 'Cholesterol'].mean()
ch_mean=ch_mean.round(2)
ch_mean
```

```python
df['Cholesterol']=df['Cholesterol'].replace(0, ch_mean)
```

```python
bp_mean = df.loc[df['RestingBP'] !=0, 'RestingBP'].mean()
bp_mean=ch_mean.round(2)
bp_mean
```

```python
df['RestingBP']=df['RestingBP'].replace(0, bp_mean)
```
