```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
arr1 = np.linspace(0, 4, 8)
arr1
```

```python
arr2 = arr1**2
arr2
```

```python
plt.plot(arr1, arr2)
plt.title("helo")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

```python
plt.subplot(3, 3, 1)
plt.plot(arr1, arr2)
plt.subplot(3, 3, 2)
plt.plot(arr1**2, arr2**2)
plt.subplot(3, 3, 3)
plt.subplot(3, 3, 4)
plt.subplot(3, 3, 5)
```

```python
fig = plt.figure(figsize=(20,8), dpi=100)
axis1= fig.add_axes([0.2,0.2,0.98,0.98])
axis1.plot(arr1, arr2)
axis2= fig.add_axes([0.4,0.4, 0.2,0.2])
axis2.plot(arr1**2, arr2)
```
